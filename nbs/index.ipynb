## Usage

### Installation

Install latest from the GitHub [repository][repo]:

```sh
$ pip install git+https://github.com/rleyvasal/sslides.git
```

or from [conda][conda]

```sh
$ conda install -c rleyvasal sslides
```

or from [pypi][pypi]


```sh
$ pip install sslides
```
